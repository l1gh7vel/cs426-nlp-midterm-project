```python
import pandas as pd
```

```python
hans_data_path = 'hans_data.csv'
midterm_bert_23_path = 'midterm_bert_23.csv'
midterm_bert_24_path = 'midterm_bert_24.csv'
midterm_bert_25_path = 'midterm_bert_25.csv'
midterm_bert_26_path = 'midterm_bert_26.csv'
midterm_bert_27_path = 'midterm_bert_27.csv'
```

```python
hans_data = pd.read_csv(hans_data_path)

midterm_bert_23 = pd.read_csv(midterm_bert_23_path)
midterm_bert_24 = pd.read_csv(midterm_bert_24_path)
midterm_bert_25 = pd.read_csv(midterm_bert_25_path)
midterm_bert_26 = pd.read_csv(midterm_bert_26_path)
midterm_bert_27 = pd.read_csv(midterm_bert_27_path)
```

```python
merged_bert_23 = pd.merge(midterm_bert_23, hans_data[['textid', 'heuristic', 'subcase']], on='textid', how='left')
merged_bert_24 = pd.merge(midterm_bert_24, hans_data[['textid', 'heuristic', 'subcase']], on='textid', how='left')
merged_bert_25 = pd.merge(midterm_bert_25, hans_data[['textid', 'heuristic', 'subcase']], on='textid', how='left')
merged_bert_26 = pd.merge(midterm_bert_26, hans_data[['textid', 'heuristic', 'subcase']], on='textid', how='left')
merged_bert_27 = pd.merge(midterm_bert_27, hans_data[['textid', 'heuristic', 'subcase']], on='textid', how='left')

merged_bert_23_cleaned = merged_bert_23.loc[:, ~merged_bert_23.columns.str.contains('^Unnamed')]
merged_bert_24_cleaned = merged_bert_24.loc[:, ~merged_bert_24.columns.str.contains('^Unnamed')]
merged_bert_25_cleaned = merged_bert_25.loc[:, ~merged_bert_25.columns.str.contains('^Unnamed')]
merged_bert_26_cleaned = merged_bert_26.loc[:, ~merged_bert_26.columns.str.contains('^Unnamed')]
merged_bert_27_cleaned = merged_bert_27.loc[:, ~merged_bert_27.columns.str.contains('^Unnamed')]
```

```python
total_rows_bert_23 = merged_bert_23_cleaned.shape[0]
print(f"Total number of rows in BERT 23: {total_rows_bert_23}")

heuristic_counts_bert_23 = merged_bert_23_cleaned['heuristic'].value_counts()
print(f"Heuristic counts in BERT 23:\n{heuristic_counts_bert_23}")
```

```python
def add_is_consistent_column(df):
    df['is_consistent'] = df.apply(lambda row: 'yes' if (row['target'] == 'entailment' and (row['heuristic'] == 'subsequence' or row['heuristic'] == 'constituent' or row['heuristic'] == 'lexical_overlap')) else 'no', axis=1)
    return df
```

```python
merged_bert_23_cleaned = add_is_consistent_column(merged_bert_23_cleaned)
merged_bert_24_cleaned = add_is_consistent_column(merged_bert_24_cleaned)
merged_bert_25_cleaned = add_is_consistent_column(merged_bert_25_cleaned)
merged_bert_26_cleaned = add_is_consistent_column(merged_bert_26_cleaned)
merged_bert_27_cleaned = add_is_consistent_column(merged_bert_27_cleaned)
```

```python
def split_consistent_non_consistent(df):
    consistent_df = df[df['is_consistent'] == 'yes']
    non_consistent_df = df[df['is_consistent'] == 'no']
    return consistent_df, non_consistent_df
```

```python
def split_by_specific_heuristic(df):
    heuristics = df['heuristic'].unique()
    heuristic_dfs = {}
    
    for heuristic in heuristics:
        if heuristic == 'lexical_overlap':
            heuristic_dfs['lexical_overlap'] = df[df['heuristic'] == heuristic]
        elif heuristic == 'subsequence':
            heuristic_dfs['subsequence'] = df[df['heuristic'] == heuristic]
        elif heuristic == 'constituent':
            heuristic_dfs['constituent'] = df[df['heuristic'] == heuristic]
    return heuristic_dfs
```

```python
consistent_bert_23_lexical_overlap_subsequence_constituent = split_by_specific_heuristic(consistent_bert_23)
non_consistent_bert_23_lexical_overlap_subsequence_constituent = split_by_specific_heuristic(non_consistent_bert_23)

consistent_bert_24_lexical_overlap_subsequence_constituent = split_by_specific_heuristic(consistent_bert_24)
non_consistent_bert_24_lexical_overlap_subsequence_constituent = split_by_specific_heuristic(non_consistent_bert_24)

consistent_bert_25_lexical_overlap_subsequence_constituent = split_by_specific_heuristic(consistent_bert_25)
non_consistent_bert_25_lexical_overlap_subsequence_constituent = split_by_specific_heuristic(non_consistent_bert_25)

consistent_bert_26_lexical_overlap_subsequence_constituent = split_by_specific_heuristic(consistent_bert_26)
non_consistent_bert_26_lexical_overlap_subsequence_constituent = split_by_specific_heuristic(non_consistent_bert_26)

consistent_bert_27_lexical_overlap_subsequence_constituent = split_by_specific_heuristic(consistent_bert_27)
non_consistent_bert_27_lexical_overlap_subsequence_constituent = split_by_specific_heuristic(non_consistent_bert_27)
```

```python
def calculate_accuracy(df):
    correct_predictions = (df['encoded_predicted'] == df['target']).sum()
    total_rows = df.shape[0]
    accuracy = correct_predictions / total_rows if total_rows > 0 else 0
    return accuracy
```
